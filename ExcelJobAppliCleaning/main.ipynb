```python
clean_df = df.copy()
```

```python
clean_df["Applied on"] = clean_df["Applied on"].str.replace(
    r"(\d)([A-Za-z])",
    r"\1 \2",
    regex=True
)
```

```python
clean_df["Applied on"].unique()
```

```python
clean_df["Applied on"] = clean_df["Applied on"].str.replace(
    "June",
    "Jun",
    regex=False
)
```

```python
clean_df["Applied on"] = pd.to_datetime(
    clean_df["Applied on"],
    format="%d %b",
    errors="coerce"
)
```
